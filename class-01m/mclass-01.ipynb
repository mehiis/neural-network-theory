<h2>Transpose/Transponointi</h2>

$A^T(i, j) = A(j,i)$
- Ei vaan käänny 90-astetta!
- $D$ =  [2, 1, 2], [4, 5, 3] <b>(2x3)</b>
- $D^T$ = [2, 4], [1, 5], [2, 3] <b>(3x2) </b>
- $(A+B)^T = A^T + B^T$
- $(AB)^T = A^T * B^T$, tässä voi tulla onglema jos A ja B ovat eri dimensioita!
